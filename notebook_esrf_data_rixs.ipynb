```python
%matplotlib widget

import os
import sys
import matplotlib.pyplot as plt

PATH = r"C:\Users\leona\OneDrive\Desktop\GitHub\RIXS_data_extraction"
sys.path.append(PATH)

from extract_rixs_spectra import ESRF_run_spectrum
from one_d_rixs_spectra import RIXS_Spectra

"""
This code processes RIXS spectra from a specified folder and a list of runs using the esrf_run_spectrum class.

Parameters:
-   folder (str): The path to the folder containing the data files.
-   runs (list): A list of run numbers to process.
-   points (list): A list of points to extract from each run. 
    Each element of the list is a list of points to extract from the corresponding run. 
    If 'all' is specified, all points are extracted.
-   x_name (str): The name of the x-axis variable.
-   y_name (str): The name of the y-axis variable.
-   norm_name (str): The name of the normalization variable.
-   motor_names (list): A list of motor names to include in the processing.

The esrf_run_spectrum object is created with the specified folder and runs. The process_spectra method is then called on this object with the specified parameters to process the spectra and return the results as an xarray Dataset.
"""



sample_name = "SCO_DSO"

# folder = r"E:\HC6552\20260708\PROCESSED_DATA\SCO\SCO_align\andor1"
# filename_packaged = "DSO_Emap_H0p20"
# runs_list = [[142],[143],[144],[145],[146],[147],[148],[149],[150],[151],[152],[153],[154], [155]]
# scans = [['all']]
# order_by_parameter = "energy"

folder = r"C:\Users\leona\OneDrive - Universität Zürich UZH\LQMR group - HC-6552\data_raw\SCO_DSO_dispersion\andor1"
# filename_packaged = "runs_4_17_DSO_for_claude"
filename_packaged = "DSO_H_dispersion"
scans = 'all'
# runs_list = [[2], [4,17], [6], [8], [10], [12], [14,15]]
runs_list = [[2], [4,17]]
order_by_parameter = "H"

# folder = r"C:\Users\leona\OneDrive - Universität Zürich UZH\LQMR group - HC-6552\data_raw\SCO_STO\andor1"
# filename_packaged = "trial_STO"
# runs_list = [[3], [11], [12]]
# scans = 'all'
# order_by_parameter = "energy"

# folder = r"E:\HC6552\20260708\PROCESSED_DATA\SCO\andor1"
# filename_packaged = "STO_dispersion_energy_576.75eV"
# runs_list = [[18], [20], [22], [24], [26], [28]]
# scans = [['all']]*len(runs_list)


x_name = 'Pixel'
y_name = 'SPC'
norm_name = 'mir_rixs'

#metadata to extract from spec files. Keys (first value in the dictionary): name after extraction.
#Value (second value in the dictionary): name in the spec file.
motor_names_spec = {
        "H": "H",
        "K": "K",
        "L": "L",
        "th": "th",
        "chi": "chi",
        "phi": "phi",
        "tth": "rtth",
        "energy": "energy",
        "x": "xsam",
        "y": "ysam",
        "z": "zsam",
        "T": "ls332_rixs_tB_sam",
        "calibration (meV/pixel)": "energy calibration (meV/px)",
        "slope": "slope"
    }

#metadata to save in csv files.
#Keys (first value in the dictionary): name in the saved file.
#Values (second value in the dictionary): name in the python variable.
motor_names_to_save_csv = {
        "H": "H",
        "K": "K",
        "L": "L",
        "th": "th",
        "chi": "chi",
        "phi": "phi",
        "tth": "tth",
        "energy": "energy",
        "x": "x",
        "y": "y",
        "z": "z",
        "T": "T",
        "mirror": "mirror",
        "filename":"filename"
    }
metadata_to_save_hdf5 = ['H', 'K', 'L','th','chi','phi','tth','energy','polarization','mirror','T','x','y','z',
                        'x_name','y_name', 'norm_name', 'filename', 'calibration', 'run', 'log']

scans_from_same_run = True      #don't change
calibration = 0.008492  
alignment_params = {"aligning_range": None,     #None: automatic
                            "fit_shifts":False, #fit shifts with a polynomial?
                            "smooth_shifts":True,   #smooth shifts with a Gaussian
                            "correlation_batch_size": 10, #pre-sum before cross-correlating
                            "poly_order": 1}
resolution_for_fitting_alignment = 0.0225
normalization_factor = 5E9   #used to divide the normalization and make it a number close to 1.

folder_save = r"C:\Users\leona\OneDrive - Universität Zürich UZH\LQMR group - HC-6552\data_extraction_trial"
filepaths_save_list = []

for runs in runs_list:
    run_spec = ESRF_run_spectrum(folder, runs, scans)
    rixs_spectra = run_spec.extract_1d_runs(x_name,
                                        y_name,
                                        norm_name, 
                                        motor_names=motor_names_spec,
                                        scans_from_same_run=scans_from_same_run)


    rixs_spectra.align_spectra(**alignment_params,
                            plot=True,)

    rixs_spectra.calibrate_energy(
        auto_elastic_determination=True,
        calibration=calibration,
        plot=True
    )

    # Saving
    file_path_save = os.path.join(folder_save,  f"{filename_packaged}_scan_{'_'.join(str(run) for run in runs)}_trial.hdf5")
    filepaths_save_list.append(file_path_save)
    rixs_spectra.save_to_hdf5(
        file_path_save=file_path_save,
        save_only_avg_spectrum=True,
        normalize_spectra=False,
        metadata_to_save=metadata_to_save_hdf5)
    
    plt.show()


rixs_spectra_packaged = RIXS_Spectra(
        file_list=filepaths_save_list,
        order_by_parameter=order_by_parameter
    )
rixs_spectra_packaged.align_spectra(method='fitting',
                                    resolution=resolution_for_fitting_alignment,
                                    plot=True
                                    )

# saving
filepath_packaged = os.path.join(folder_save, f"{filename_packaged}.hdf5")
rixs_spectra_packaged.save_to_hdf5(
        filename=filepath_packaged,
        variable_names=['x','y','error'],
        divide_normalization_by_value=5E9,
        normalize_spectra=True,
        additional_metadata={'sample_name': sample_name}
    )
rixs_spectra_packaged.save_to_csv(filepath_packaged.replace('.hdf5', '.csv'),
                                  normalize_spectra=True,
                                  save_errorbars=False,
                                  divide_normalization_by_value=normalization_factor,
                                   motors_dict=motor_names_to_save_csv)

# # run_spec.save_to_hdf5(file_path_save)
# rixs_spectra.save_to_csv_for_originlab(file_path_save.replace('.hdf5', '.csv'),
#                                    motor_names=['srot', 'energy'],
#                                    motor_name_mapping={'theta': 'srot', 'energy': 'energy'},)
```
